```python
# Step 6: Create SQLite databases for all three medical datasets

import sqlite3
import os

# Create databases folder
os.makedirs("/content/databases", exist_ok=True)

def dataframe_to_sqlite(df, db_path, table_name):
    conn = sqlite3.connect(db_path)

    # Save DataFrame as SQLite table
    df.to_sql(
        table_name,
        conn,
        if_exists="replace",
        index=False
    )

    conn.close()
    print(f"Created: {db_path}")
    print(f"Table: {table_name}")
    print(f"Rows: {len(df)}")
    print("-" * 50)


# 1. Heart Disease Database
dataframe_to_sqlite(
    heart_df,
    "/content/databases/heart_disease.db",
    "heart_disease"
)

# 2. Cancer Database
dataframe_to_sqlite(
    cancer_df,
    "/content/databases/cancer.db",
    "cancer"
)

# 3. Diabetes Database
dataframe_to_sqlite(
    diabetes_df,
    "/content/databases/diabetes.db",
    "diabetes"
)

print("All three SQLite databases created successfully! ✅")
```

```python
from duckduckgo_search import DDGS

def free_web_search(query, max_results=5):
    results = []

    with DDGS() as ddgs:
        search_results = ddgs.text(
            query,
            max_results=max_results
        )

        for item in search_results:
            results.append({
                "title": item.get("title", ""),
                "body": item.get("body", ""),
                "href": item.get("href", "")
            })

    return results

test_results = free_web_search(
    "common symptoms of diabetes",
    max_results=3
)

for i, result in enumerate(test_results, 1):
    print(f"\nResult {i}")
    print("Title:", result["title"])
    print("Summary:", result["body"])
    print("URL:", result["href"])
```

```python
from ddgs import DDGS

def free_web_search(query, max_results=5):
    results = []

    with DDGS() as ddgs:
        search_results = ddgs.text(
            query,
            max_results=max_results
        )

        for item in search_results:
            results.append({
                "title": item.get("title", ""),
                "body": item.get("body", ""),
                "href": item.get("href", "")
            })

    return results

test_results = free_web_search(
    "common symptoms of diabetes",
    max_results=3
)

print("Number of results:", len(test_results))

for i, result in enumerate(test_results, 1):
    print(f"\nResult {i}")
    print("Title:", result["title"])
    print("Summary:", result["body"])
    print("URL:", result["href"])
```

```python
from langchain_core.tools import tool

@tool
def MedicalWebSearchTool(query: str) -> str:
    """
    Search the web for general medical information such as disease definitions,
    symptoms, causes, prevention, treatment information, and health guidance.
    This tool does not provide personal diagnosis.
    """
    try:
        results = free_web_search(query, max_results=5)

        if not results:
            return "No relevant web results found."

        formatted_results = []

        for i, result in enumerate(results, 1):
            formatted_results.append(
                f"Result {i}:\n"
                f"Title: {result['title']}\n"
                f"Summary: {result['body']}\n"
                f"URL: {result['href']}"
            )

        return "\n\n".join(formatted_results)

    except Exception as e:
        return f"Web search failed: {str(e)}"

print("MedicalWebSearchTool created successfully!")
print("Tool name:", MedicalWebSearchTool.name)
print("Tool description:", MedicalWebSearchTool.description)
```

```python
# ============================================================
# Main Medical Routing Agent
# ============================================================

def MedicalRoutingAgent(question: str):
    """
    Main routing agent that selects the appropriate tool
    based on the user's question.
    """

    q = question.lower()

    # Heart Disease related keywords
    heart_keywords = [
        "heart", "cardiac", "cholesterol", "trestbps",
        "thalach", "chest pain", "heart disease"
    ]

    # Cancer related keywords
    cancer_keywords = [
        "cancer", "tumor", "tumour", "genetic risk",
        "cancer history", "diagnosis"
    ]

    # Diabetes related keywords
    diabetes_keywords = [
        "diabetes", "glucose", "insulin", "pregnancies",
        "blood sugar", "diabetic", "diabetes pedigree"
    ]

    # Dataset-related questions
    if any(keyword in q for keyword in heart_keywords):
        selected_tool = "HeartDiseaseDBTool"
        tool_result = HeartDiseaseDBTool.invoke({"question": question})

    elif any(keyword in q for keyword in cancer_keywords):
        selected_tool = "CancerDBTool"
        tool_result = CancerDBTool.invoke({"question": question})

    elif any(keyword in q for keyword in diabetes_keywords):
        selected_tool = "DiabetesDBTool"
        tool_result = DiabetesDBTool.invoke({"question": question})

    else:
        selected_tool = "MedicalWebSearchTool"
        tool_result = MedicalWebSearchTool.invoke({"query": question})

    return {
        "question": question,
        "selected_tool": selected_tool,
        "answer": tool_result
    }

print("Medical Routing Agent created successfully!")
```

```python
# ============================================================
# Final Improved Routing Logic
# ============================================================

def detect_selected_tool(question: str):
    q = question.lower()

    # Dataset/statistical question indicators
    dataset_keywords = [
        "dataset", "database", "records", "record",
        "average", "mean", "count", "number of",
        "total", "maximum", "minimum", "percentage",
        "how many", "distribution", "column",
        "patients in", "data"
    ]

    # General medical information keywords
    web_keywords = [
        "symptom", "symptoms",
        "cause", "causes",
        "prevention", "prevent",
        "treatment", "treat",
        "medicine", "medication",
        "definition",
        "sign", "signs",
        "risk factor", "risk factors",
        "how to prevent",
        "cure",
        "meaning of"
    ]

    # Dataset questions get priority
    is_dataset_question = any(
        keyword in q for keyword in dataset_keywords
    )

    if is_dataset_question:

        heart_keywords = [
            "heart", "cardiac", "cholesterol",
            "trestbps", "thalach",
            "chest pain", "heart disease"
        ]

        cancer_keywords = [
            "cancer", "tumor", "tumour",
            "genetic risk", "cancer history"
        ]

        diabetes_keywords = [
            "diabetes", "glucose", "insulin",
            "pregnancies", "blood sugar",
            "diabetic", "diabetes pedigree"
        ]

        if any(keyword in q for keyword in heart_keywords):
            return "HeartDiseaseDBTool"

        elif any(keyword in q for keyword in cancer_keywords):
            return "CancerDBTool"

        elif any(keyword in q for keyword in diabetes_keywords):
            return "DiabetesDBTool"

    # General medical information -> Web Search
    if any(keyword in q for keyword in web_keywords):
        return "MedicalWebSearchTool"

    # Disease name based fallback
    if "heart" in q:
        return "HeartDiseaseDBTool"

    elif "cancer" in q:
        return "CancerDBTool"

    elif "diabetes" in q or "glucose" in q:
        return "DiabetesDBTool"

    else:
        return "MedicalWebSearchTool"


test_questions = [
    "How many records are in the heart disease dataset?",
    "What is the average age in the cancer dataset?",
    "How many diabetic patients are present?",
    "What are the common symptoms of diabetes?",
    "What is cancer prevention?",
    "What is the definition of insulin?",
    "What is the average glucose level in the diabetes dataset?",
    "What are the causes of heart disease?",
    "How many cancer patients are diagnosed?"
]

print("Final Routing Test Results:\n")

for question in test_questions:
    print("Question:", question)
    print("Selected Tool:", detect_selected_tool(question))
    print("-" * 70)
```
